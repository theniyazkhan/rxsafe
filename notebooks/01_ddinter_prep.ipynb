```python
import glob
import pandas as pd

files = glob.glob("../data/external/ddinter_2026-09-26/ddinter_downloads_code_*.csv")
df_list = [pd.read_csv(f) for f in files]
master_df = pd.concat(df_list, ignore_index=True).drop_duplicates()
master_df.to_csv("../data/external/ddinter_2026-09-26/ddinter_combined.csv", index=False)
print(f"Total interaction pairs: {len(master_df)}")
```
